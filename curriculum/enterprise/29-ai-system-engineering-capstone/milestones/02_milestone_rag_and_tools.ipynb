# Milestone 2: Context & Action

## Goal
The copilot is useless without data. Connect the Copilot to a simulated Knowledge Base (RAG) and give it a `check_order_status` tool to execute against a simulated database.

```python
from google import genai
from google.genai import types

client = genai.Client()
MODEL_ID = 'gemini-2.5-flash'

def check_order_status(order_id: str) -> str:
    """Simulates a database lookup for an order."""
    orders = {"ORD123": "Shipped", "ORD456": "Pending", "ORD789": "Cancelled"}
    return orders.get(order_id, "Order not found")

KNOWLEDGE_BASE = """
Return Policy: Customers can return items within 30 days of shipment. 
Shipping Policy: Standard shipping takes 3-5 business days. Expedited takes 1-2 days.
"""
```

```python
def handle_query_with_context(query: str) -> str:
    prompt = f"""
    You are a support agent. Answer the user's query using the provided knowledge base.
    If the user asks about an order, use the provided tool.
    
    <knowledge_base>
    {KNOWLEDGE_BASE}
    </knowledge_base>
    
    User Query: {query}
    """
    
    response = client.models.generate_content(
        model=MODEL_ID,
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=0.0,
            tools=[check_order_status]
        )
    )
    return response.text

print(handle_query_with_context("What is your return policy?"))
```
