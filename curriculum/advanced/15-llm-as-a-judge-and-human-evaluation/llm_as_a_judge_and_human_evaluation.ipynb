Judge outputs require validation against human and deterministic signals.

```python
from importlib.util import module_from_spec,spec_from_file_location
from pathlib import Path
import sys
p=Path.cwd()/'curriculum/advanced/15-llm-as-a-judge-and-human-evaluation/lab.py'
if not p.exists():p=Path.cwd()/'lab.py'
s=spec_from_file_location('judge_lab',p);lab=module_from_spec(s);sys.modules[s.name]=lab;s.loader.exec_module(lab)
```

## Agreement and bias experiment

Agreement is a diagnostic, not proof. Pairwise order must be randomized to avoid position bias.

```python
score=lab.agreement(lab.HUMAN,lab.JUDGE)
assert score<1
{'agreement':score,'order_a_b':lab.pairwise(3,2),'order_b_a':lab.pairwise(2,3)}
```
