# 19 — Prompting for Coding Agents

A coding task needs acceptance criteria and verification, not a vague request.

```python
from importlib.util import module_from_spec,spec_from_file_location
from pathlib import Path
import sys
p=Path.cwd()/'curriculum/advanced/19-prompting-for-coding-agents/lab.py'
if not p.exists():p=Path.cwd()/'lab.py'
s=spec_from_file_location('code_lab',p);lab=module_from_spec(s);sys.modules[s.name]=lab;s.loader.exec_module(lab)
```

```python
vague=lab.assess({'problem':'fix authentication'})
structured=lab.assess({'problem':'expired sessions fail','scope':'auth middleware','tests':'pytest tests/test_auth.py','completion':'tests pass and diff reviewed'})
assert not vague['complete'] and structured['complete']
{'vague':vague,'structured':structured}
```

Production: require repository inspection, file selection, plan, tests, diff review, and rollback evidence.
