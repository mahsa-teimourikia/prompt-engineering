# 08 — Context Engineering

Compare full context with selected authorized evidence. This lab is offline and does not call a model.

```python
from importlib.util import module_from_spec,spec_from_file_location
from pathlib import Path
import sys
p=Path.cwd()/'curriculum/intermediate/08-context-engineering/lab.py'
if not p.exists(): p=Path.cwd()/'lab.py'
s=spec_from_file_location('context_lab',p); lab=module_from_spec(s);sys.modules[s.name]=lab;s.loader.exec_module(lab)
```

## Experiment

The full packet includes irrelevant, untrusted content. Selection retains only relevant high-authority evidence.

```python
full=lab.assemble('full'); selected=lab.assemble('selected')
assert not full['trusted'] and selected['trusted']
{'full':full,'selected':selected}
```
