# 13 — Prompt Security and Untrusted Content

Attack a vulnerable refund workflow, then retest with deterministic application boundaries. Offline only.

```python
from importlib.util import module_from_spec,spec_from_file_location
from pathlib import Path
import sys
p=Path.cwd()/'curriculum/intermediate/13-prompt-security-and-untrusted-content/lab.py'
if not p.exists():p=Path.cwd()/'lab.py'
s=spec_from_file_location('security_lab',p);lab=module_from_spec(s);sys.modules[s.name]=lab;s.loader.exec_module(lab)
```

## Attack and trace

The vulnerable design lets untrusted content drive an unauthorized effect and exfiltration.

```python
attack=lab.vulnerable(lab.ATTACK)
assert attack['exfiltrates']
attack
```

## Deterministic boundary and retest

The defended path does not execute or release data. Prompt defenses reduce risk; application controls enforce security.

```python
defended=lab.defended(lab.ATTACK)
assert not defended['exfiltrates'] and defended['action']=='draft_or_escalate'
defended
```
