# 11 — Tool Calling and Tool Interface Design

Tool selection is a model proposal; authorization is deterministic application policy.

```python
from importlib.util import module_from_spec,spec_from_file_location
from pathlib import Path
import sys
p=Path.cwd()/'curriculum/intermediate/11-tool-calling-and-tool-interface-design/lab.py'
if not p.exists():p=Path.cwd()/'lab.py'
s=spec_from_file_location('tool_lab',p);lab=module_from_spec(s);sys.modules[s.name]=lab;s.loader.exec_module(lab)
```

## Experiment

Compare unknown, incomplete, and valid tool proposals. A valid proposal is still not permission to act.

```python
assert lab.select('delete_order',{})[0] is False
assert lab.select('get_order_status',{})[0] is False
assert lab.select('draft_refund',{'order_id':'42','reason':'damaged'})[0] is True
[lab.select('delete_order',{}),lab.select('get_order_status',{}),lab.select('draft_refund',{'order_id':'42','reason':'damaged'})]
```
